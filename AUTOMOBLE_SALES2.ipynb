```python
# We will be seeing if there are any null values present or not
dataset.isnull().sum()
# Since there are no null values which means out dataset is complete
```

```python
# based on the dataset now we will be separating the month day and year
dataset['DAY'] = dataset['ORDERDATE'].str.split('/').str[0]
dataset['MONTH'] = dataset['ORDERDATE'].str.split('/').str[1]
dataset['YEAR'] = dataset['ORDERDATE'].str.split('/').str[2]
dataset
```

```python
# Since the date has been spearated now we are going to remove the order date
dataset = dataset.drop('ORDERDATE', axis =1)
dataset
```

```python
# We will be seeing the type of data for date and if it is object we will convert to integer
dataset
print(dataset['DAY'].dtype)
```

```python
# Since the day, month nd year are object now we convert them into integers
dataset['DAY'] = dataset['DAY'].astype(int)
dataset['MONTH'] = dataset['MONTH'].astype(int)
dataset['YEAR'] = dataset['YEAR'].astype(int)
```

```python
# Some basic questions for more exploration
```

```python
# 1) Which country's sells the most and least automobiles?
max_sales = dataset['COUNTRY'].value_counts()
print(max_sales)
# USA sells the maximum automobilies
# Ireland is the lowest country for sales
```

```python
# 2) What is the most and least sold product line?
most_product_line = dataset['PRODUCTLINE'].value_counts()
print(most_product_line)
# Classic cars are the most sold followed by Vintage Cars and Motorcycles.
# Trains are the least sold
```

```python
# 3) Top and less top cities for automobiles
top_cities = dataset['CITY'].value_counts()
print(top_cities)
# Madrid is the top city for sales of automobilies followed by San Rafael and NYC
# Charleroi is the least city for sales
```

```python
# 4) Highest and lowest sales
highest_sales = dataset['SALES'].max()
print(highest_sales)
highest_sales_row = dataset[dataset['SALES']==highest_sales]
highest_sales_row
# The highest sales was 14082.8 which was for vintage cars made by Sue fro USA at 2020
```

The row containing the max sales has DEAL_SIZE = Large in city of San Jose of COUNTRY USA

```python
lowest_sales = dataset['SALES'].min()
print(lowest_sales)
lowest_sales_row = dataset[dataset['SALES']==lowest_sales]
lowest_sales_row
# Lowest sale was made on France at year 2020 by Janine for Trucks and Buses
```

```python
# 5) Highest and lowest dealsize
highest_deal_size = dataset['DEALSIZE'].value_counts()
print(highest_deal_size)
# The highes deal size is Medium and lowest deal size is large
```

```python
# 6) Which year were the automobiles sold the most and  least?
most_year = dataset['YEAR'].value_counts()
print(most_year)
# In the year 2019 most automobiles were sold
# In the year 2020 least automobilies were sold. It may be due to less data.
```

```python
# 7) HIghest and lowest price of automobilies.
highest_price = dataset['PRICEEACH'].max()
print(highest_price)
highest_price_row = dataset[dataset['PRICEEACH']==highest_price]
print(highest_price_row)
# Vintage cars has highest price
```

```python
lowest_price = dataset['PRICEEACH'].min()
print(lowest_price)
lowest_price_row = dataset[dataset['PRICEEACH']==lowest_price]
print(lowest_price_row)
# Vintage Cars has lowest price cars has lowest price
```

Here the min and max price are for the same Vintage Cars. It may be because the Vintage Cars may be further divided into the other expensive and cheap types. Another reason may also be due to the size of the deal.


```python
# 8) Average money spent by people on each country?
avg_spent = dataset.groupby('COUNTRY')['SALES'].mean().sort_values(ascending = False)
print(avg_spent)
# Denmark has highest average spent and Canada has the lowest
```

From the output, it is seen that the sales is minimun in Canada and maximum in Denmark

```python
# 9) Average in terms of product?
avg_product = dataset.groupby('PRODUCTLINE')['SALES'].mean().sort_values(ascending = False)
print(avg_product)
# Classic cars are sold most in average and Trains are less
```

From the output, it is seen that the sales is minimun through trains and maximum through classic cars

```python
# Distribution of products

item_count = dataset['PRODUCTLINE'].value_counts()
plt.figure(figsize = (8,8))
pie_chart = plt.pie(item_count,labels=item_count.index,autopct = '%1.2f%%')
plt.title('Distribution of Product Lines')
plt.show()
```

```python
# Classic cars has the highest distribution of 34.5% followed by Vintage Cars and Motorcycles. Trains has the lowest
# followed by ships and trucks.
```

```python
# Distribution by country
```

```python
country_counts = dataset['COUNTRY'].value_counts()
country_counts
```

```python
# country_counts = dataset['COUNTRY'].value_counts()
plt.figure(figsize=(18, 6))
sns.countplot(x=dataset['COUNTRY'])
plt.title('Number of Orders by Country')
plt.xlabel('Country')
plt.ylabel('Number of Orders')
plt.show()
# USA has highest number of orders
```

```python
country_sales = dataset.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False)
country_sales
```

```python
# Sales made
country_sales = dataset.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False)
plt.figure(figsize = (12,8))
h_bar= plt.bar(country_sales.index, country_sales.values)
plt.xlabel('Country')
plt.ylabel('Sales')
plt.title('Sales by Country (Descending Order)')
plt.xticks(rotation=90)
plt.show()
# USA has the highest number of total sales made whereas Ireland has the lowest
```

```python
# For dealsize
item_counts = dataset['DEALSIZE'].value_counts()
plt.figure(figsize=(10,5))
pie_chart = plt.pie(item_counts,labels= item_counts.index,autopct = '%1.1f%%',startangle = 90)
plt.title('Distribution of Deal Sizes')
plt.show()
# Medium and small sizes are similar with 49.1% and 45.1%
# Large has the lowest with 5.5%
```

The figure says that the twp parties are dealing with medium range most of the time and dealing with large range rarely.

```python
# Status
item_count = dataset['STATUS'].value_counts()
plt.figure(figsize = (10,5))
bar_graph = plt.bar(item_count.index, item_count.values)
plt.xlabel('Status')
plt.ylabel('Number of Orders')
plt.title('Number of Orders by Status')
plt.show()
# More than 2000 orders have been shipped
```
